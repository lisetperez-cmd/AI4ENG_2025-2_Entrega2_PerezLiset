```python
df = pd.read_csv('train.csv')
df
```

```python
#convertimos los valores de la columna E_VALORMATRICULAUNIVERSIDAD a valores numéricos, para que nos dé el promedio de los mismos

def convertir_valor_matricula(valor):
    if pd.isna(valor):
        return np.nan
    elif 'Menos de 500 mil' in valor:
        return 250000  # Asignamos un valor promedio entre 0 y 500 mil
    elif 'Entre 500 mil y menos de 1 millón' in valor:
        return 750000  # Promedio entre 500 mil y 1 millón
    elif 'Entre 1 millón y menos de 2.5 millones' in valor:
        return 1750000  # Promedio entre 1 y 2.5 millones
    elif 'Entre 2.5 millones y menos de 4 millones' in valor:
        return 3250000  # Promedio entre 2.5 y 4 millones
    elif 'Entre 4 millones y menos de 5.5 millones' in valor:
        return 4750000  # Promedio entre 4 y 5.5 millones
    elif 'Entre 5.5 millones y menos de 7 millones' in valor:
        return 6250000  # Promedio entre 5.5 y 7 millones
    elif 'Más de 7 millones' in valor:
        return 7500000  # Asignamos un valor mínimo representativo superior a 7 millones
    elif 'No pagó matrícula' in valor:
        return 0  # Asumimos que no se pagó nada
    else:
        return np.nan  # Para cualquier caso que no coincida
```

```python
df_copy = df.copy()
df_copy['E_VALORMATRICULAUNIVERSIDAD'] = df_copy['E_VALORMATRICULAUNIVERSIDAD'].apply(convertir_valor_matricula)

media = df_copy ['E_VALORMATRICULAUNIVERSIDAD'].mean()
df_copy['E_VALORMATRICULAUNIVERSIDAD'] = df_copy['E_VALORMATRICULAUNIVERSIDAD'].fillna(media.round(3))

df_copy['E_VALORMATRICULAUNIVERSIDAD'].unique() # Se verifica que ya no queden valores nulos en la columna E_VALORMATRICULAUNIVERSIDAD
```

# Se define la función para convertir los valores de ESTU_HORASSEMANATRABAJA en datos numéricos.

```python
#convertimos los valores de la columna E_HORASSEMANATRABAJA a valores numéricos,
def convertir_horas_trabajadas(valor):
    if isinstance(valor, str):
        if "Entre" in valor:
            partes = valor.split('y')
            min_val = float(partes[0].split(' ')[-2])  # Obtener el penúltimo elemento
            max_val = float(partes[1].split(' ')[-2])  # Obtener el penúltimo elemento de la segunda parte
            return (min_val + max_val) / 2  # Retornar el promedio del rango

        elif "Más de" in valor:
            return float(valor.split(' ')[2])  # Convertir a número

        elif "Menos de" in valor:
            return float(valor.split(' ')[2])  # Convertir a número

        elif "0" in valor:
            return 0  # Devolver 0 en número

    return np.nan  # Devolver NaN si no es un valor válido
```

```python
# Applicamos la conversión y eliminamos los valores nulos para la columna E_HORASSEMANATRABAJA

df_copy['E_HORASSEMANATRABAJA'] = df_copy['E_HORASSEMANATRABAJA'].apply(convertir_horas_trabajadas)

df_copy['E_HORASSEMANATRABAJA'] = df_copy['E_HORASSEMANATRABAJA'].fillna(0)

df_copy['E_HORASSEMANATRABAJA'].unique() # Se verifica que ya no queden valores nulos en la columna ESTU_HORASSEMANATRABAJA
```

```python
#Creamos un diccionario para la correspondiente F_ESTRATOVIVIENDA y se hacen los reemplazos necesarios

estrato_dict = {
    'Estrato 1': 1,
    'Estrato 2': 2,
    'Estrato 3': 3,
    'Estrato 4': 4,
    'Estrato 5': 5,
    'Estrato 6': 6,
    'Sin Estrato': 0,
    np.nan: 0  # Cambiar nan a 0
}

df_copy['F_ESTRATOVIVIENDA'] = df_copy['F_ESTRATOVIVIENDA'].replace(estrato_dict)

df_copy['F_ESTRATOVIVIENDA'].unique() # Se verifica que ya no queden valores nulos en la columna FAMI_ESTRATOVIVIENDA
```

```python
# Comenzamos con la conversión a binomiales, anulamos los valores nulos para E_PAGOMATRICULAPROPIO
df_copy['E_PAGOMATRICULAPROPIO'] = df_copy['E_PAGOMATRICULAPROPIO'].replace({'Si': 1, 'No': 0})

df_copy['E_PAGOMATRICULAPROPIO'] = df_copy['E_PAGOMATRICULAPROPIO'].fillna(0)

df_copy['E_PAGOMATRICULAPROPIO'].unique() # Se verifica que ya no queden valores nulos en la columna E_PAGOMATRICULAPROPIO
```

```python
#Usamos un segundo diccionario para la columna RENDIMIENTO_GLOBAL que es nuestro objetivo, asignando la clasificiación en orden de aumento

rend_dict = {
    'bajo': 1,
    'medio-bajo': 2,
    'medio-alto': 3,
    'alto': 4,
}

df_copy['RENDIMIENTO_GLOBAL'] = df_copy['RENDIMIENTO_GLOBAL'].replace(rend_dict)

df_copy['RENDIMIENTO_GLOBAL'].unique() # Se verifica que ya no queden valores nulos en la columna RENDIMIENTO_GLOBAL
```

```python
df_copy['F_TIENEINTERNET.1'] = df_copy['F_TIENEINTERNET'].replace({'Si': 1, 'No': 0})

df_copy['F_TIENEINTERNET.1'] = df_copy['F_TIENEINTERNET.1'].fillna(0)

df_copy = df_copy.drop('F_TIENEINTERNET', axis=1) # Drop the original string column

df_copy['F_TIENEINTERNET.1'].unique() # Se verifica que ya no queden valores nulos en la columna F_TIENEINTERNET.1
```

```python
def to_onehot(x):
    values = np.unique(x)
    r = np.r_[[np.argwhere(i==values)[0][0] for i in x]]
    return np.eye(len(values))[r].astype(int)

def replace_column_with_onehot(d, col):
    assert sum(d[col].isna())==0, "column must have no NaN values"
    values = np.unique(d[col]
                      )
    k = to_onehot(d[col].values)
    r = pd.DataFrame(k, columns=["%s_%s"%(col, values[i]) for i in range(k.shape[1])], index=d.index).join(d)
    del(r[col])
    return r
```

```python
col_m = 'F_EDUCACIONMADRE'
df_copy[col_m].fillna('No Aplica', inplace=True)
madre_onehot = replace_column_with_onehot(df_copy[[col_m]], col_m)
df_copy = df_copy.join(madre_onehot)
df_copy = df_copy.drop(col_m, axis=1)

col_p = 'F_EDUCACIONPADRE'
df_copy[col_p].fillna('No Aplica', inplace=True)
padre_onehot = replace_column_with_onehot(df_copy[[col_p]], col_p)
df_copy = df_copy.join(padre_onehot)
df_copy = df_copy.drop(col_p, axis=1)

df_copy
```

```python
# Eliminamos dos columnas
df_copy = df_copy.drop(['E_PRGM_ACADEMICO'], axis=1, errors='ignore')
df_copy = df_copy.drop(['PERIODO_ACADEMICO'], axis=1, errors='ignore')
df_copy = df_copy.drop(['E_PRGM_DEPARTAMENTO'], axis=1, errors='ignore')

df_copy
```

```python
X_train = df_copy.drop(['RENDIMIENTO_GLOBAL', 'ID'], axis=1)
y_train = df_copy['RENDIMIENTO_GLOBAL']
```

```python
#Entrenamiento para RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier

# Convert remaining string columns to numerical
# Assuming 'Si': 1, 'No': 0 for boolean-like features
boolean_cols_to_convert = ['F_TIENELAVADORA', 'F_TIENEAUTOMOVIL', 'F_TIENECOMPUTADOR'] # Corrected typo here
for col in boolean_cols_to_convert:
    if col in X_train.columns:
        X_train[col] = X_train[col].replace({'Si': 1, 'No': 0}).fillna(0)

# Convert 'E_PRIVADO_LIBERTAD' (assuming 'N': 0, 'S': 1)
if 'E_PRIVADO_LIBERTAD' in X_train.columns:
    X_train['E_PRIVADO_LIBERTAD'] = X_train['E_PRIVADO_LIBERTAD'].replace({'N': 0, 'S': 1}).fillna(0)

model = Pipeline(steps=[('clasifier', RandomForestClassifier(random_state=42))])
model.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score

y_val_pred = model.predict(X_train)
accuracy = accuracy_score(y_train, y_val_pred)
print("Accuracy:", accuracy)
```
